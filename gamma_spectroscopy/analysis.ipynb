```python
# Uncomment the next line to make the plot interactive (may require installing the `ipympl` package):
# %matplotlib widget

import matplotlib.pyplot as plt
import numpy as np

# Load the NaI and HPGe environmental background spectra
data = np.loadtxt('data/background-NaI.txt')
data02 = np.loadtxt('data/background-HPGe.txt')

# Append new columns to data and data02
# data = np.column_stack((data, data[:, 0]*3.506685417-10.16497202))
data02 = np.column_stack((data02, data02[:, 0]*0.23636))

# Plot the data
plt.figure(figsize=(10, 6))
# plt.plot(data[:, 2], data[:, 1])
plt.plot(data02[:, 2], data02[:, 1])
plt.yscale('log')
#plt.ylim(0, 10**5)

# Add more minor tick subdivisions to the axis
plt.minorticks_on()

plt.xlabel('Energy (keV)')
# plt.xlabel('Channel')  # There used to be two X labels here; kept this one active
plt.ylabel('Counts')
plt.title('Environmental Background Measured with the HPGe Detector')

# Show both major and minor gridlines
plt.grid(True, which='major', linestyle='-', linewidth=0.8)
plt.grid(True, which='minor', linestyle=':', alpha=0.5)

plt.show()
```

```python
# Load the Mg4 sample spectrum and its environmental background
data = np.loadtxt('data/Mg4-NaI.txt')
data02 = np.loadtxt('data/background-NaI.txt')

# Normalize both spectra by live time, then subtract background from sample
data = np.column_stack((data, data[:, 1]/5930))
data02 = np.column_stack((data02, data02[:, 1]/7999))
data = np.column_stack((data, data[:, 2]-data02[:, 2]))

# Save the background-subtracted spectrum as a tab-separated txt file
np.savetxt('data/Mg4-NaI-background-subtracted.txt', np.column_stack((data[:, 0], data[:, 3])), delimiter='\t', fmt='%g')

# Plot the data
plt.figure(figsize=(10, 6))
plt.plot(data[:, 0], data[:, 3])
#plt.plot(data02[:, 2], data02[:, 1])
#plt.yscale('log')
#plt.ylim(0, 10**5)

# Add more minor tick subdivisions to the axis
plt.minorticks_on()

plt.xlabel('Channel')
# plt.xlabel('Channel')  # There used to be two X labels here; kept this one active
plt.ylabel('Counts')
plt.title('Mg4 Sample - NaI Detector, Background Subtracted')

# Show both major and minor gridlines
plt.grid(True, which='major', linestyle='-', linewidth=0.8)
plt.grid(True, which='minor', linestyle=':', alpha=0.5)

plt.show()
```
