```python
# Uncomment the next line to make the plot interactive (may require installing the `ipympl` package):
# %matplotlib widget

import matplotlib.pyplot as plt
import numpy as np

# Load a coincidence spectrum (channel index -> counts)
data = np.loadtxt('data/90.txt')
# data02 = np.loadtxt('data/157.txt')

# Append new columns to data and data02
# data = np.column_stack((data, data[:, 0]*3.447))

# Plot the data
plt.figure(figsize=(10, 6))
plt.plot(data)#[:, 2], data[:, 1])
# plt.plot(data02)
#plt.plot(data02[:, 2], data02[:, 1])
#plt.yscale('log')
#plt.ylim(0, 10**5)

# Add more minor tick subdivisions to the axis
plt.minorticks_on()

plt.xlabel('Channel')
# plt.xlabel('Channel')  # There used to be two X labels here; kept this one active
plt.ylabel('Counts')
plt.title('Compton Coincidence Spectrum - 90 Degree Run (NaI Detectors)')

# Show both major and minor gridlines
plt.grid(True, which='major', linestyle='-', linewidth=0.8)
plt.grid(True, which='minor', linestyle=':', alpha=0.5)

plt.show()
```
